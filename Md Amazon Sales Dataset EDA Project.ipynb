#**Detecting Outliers Using IQR Method**

```python
Q1=df['price'].quantile(0.25)
Q3=df['price'].quantile(0.75)
IQR=Q3-Q1
Lowerbound=Q1-1.5*IQR
Upperbound=Q3+1.5*IQR
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Fence:", Lowerbound)
print("Upper Fence:", Upperbound)
```

####Conclusion

According to the IQR method, there are no outliers in the price column . The prices range from 5.01 to 499.99, which is completely within the calculated lower and upper fences of -244.90 and 749.08. Therefore, the price data does not contain any extreme values that would be considered outliers using the 1.5 × IQR rule.

```python
Q1 = df['total_revenue'].quantile(0.25)
Q3 = df['total_revenue'].quantile(0.75)

IQR = Q3 - Q1

lower_fence = Q1 - 1.5 * IQR
upper_fence = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Fence:", lower_fence)
print("Upper Fence:", upper_fence)
```

####Conclusion

The middle 50% of total revenue values lie between 240.32 and 968.97. The IQR is 728.65, giving a lower fence of -852.655 and an upper fence of 2061.945. Therefore, any total_revenue value above 2061.945 would be considered an outlier according to the 1.5 × IQR rule.

```python
# Step1:Outliers count
outliers = df[
    (df['total_revenue'] < lower_fence) |
    (df['total_revenue'] > upper_fence)
]

print("Number of Outliers:", len(outliers))
```

```python
#Step2:percentage of outlier
outlier_percentage = (len(outliers) / len(df)) * 100

print("Outlier Percentage:", round(outlier_percentage, 2), "%")
```

```python
# check the outliers
outliers[
    ['price', 'discount_percent', 'quantity_sold',
     'discounted_price', 'total_revenue']
].head(10)
```

#**Conclusion**

752 potential outliers were detected in total revenue using the IQR method, representing 1.5% of the dataset. After inspection, these values appear to be valid high-value transactions, mainly caused by higher product prices and quantities. Therefore, the outliers were retained for further analysis


Although 752 (1.5%) potential outliers were detected in total revenue using the IQR method, they were not removed because inspection showed that they represent valid high-value transactions rather than data errors. Therefore, these observations were retained for further analysis.

#**Feature Engineering**

```python
df['year'] = df['order_date'].dt.year
df['month'] = df['order_date'].dt.month
df['month_name'] = df['order_date'].dt.month_name()
df['day_name'] = df['order_date'].dt.day_name()

df[['order_date', 'year', 'month', 'month_name', 'day_name']].head()
```

##Conclusion

This step prepares the dataset for analysis by creating useful time-based features such as year, month, and month-year for studying sales trends. It also validates the discounted price by comparing the actual value with the expected value based on the discount percentage. This helps identify possible data errors and ensures the dataset is reliable before performing further analysis.

 Date-based feature engineering on the order_date column. It extracts the year, month number, month name, and day name and stores them as new columns. These new features make the date information easier to analyze and can help identify sales patterns by year, month, and day of the week
